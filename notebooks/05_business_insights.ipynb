# BSD-Analytics: 05. Business Insights & Fleet Rebalancing
Translating machine learning demand models into operational decisions and van dispatch schedules.

```python
from bsd_analytics.analytics.insights import generate_fleet_recommendations, get_executive_kpis
from bsd_analytics.data.ingestion import load_raw_data

df = load_raw_data()
kpis = get_executive_kpis(df)
print("Executive KPIs:", kpis)

recommendations = generate_fleet_recommendations(df)
for idx, rec in enumerate(recommendations, 1):
    print(f"\n[{idx}] {rec['category']}")
    print(f"Recommendation: {rec['recommendation']}")
    print(f"Business Impact: {rec['business_impact']}")
```
